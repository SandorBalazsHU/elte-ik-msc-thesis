```python
# Szükséges könyvtárak importálása
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Adatok megadása
samples = ['1k', '5k', '10k', '25k']
top1_acc = [0.3022, 0.4603, 0.5133, 0.5499]
top5_acc = [0.6505, 0.7766, 0.8177, 0.8376]
precision = [0.2851, 0.4254, 0.4782, 0.5310]
recall = [0.3149, 0.4762, 0.5421, 0.5978]
f1_score = [0.2820, 0.4325, 0.4905, 0.5436]
inference_time = [31.49, 31.72, 31.71, 31.82]

# Táblázat létrehozása (opcionális megjelenítéshez)
df = pd.DataFrame({
    'Samples': samples,
    'Top-1 Accuracy': top1_acc,
    'Top-5 Accuracy': top5_acc,
    'Precision': precision,
    'Recall': recall,
    'F1-score': f1_score,
    'Inference Time (s)': inference_time
})

display(df)

# Grafikonok készítése
fig, axs = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle('ResNet50 teljesítmény különböző tanítási méreteken', fontsize=16)

# Top-1 és Top-5 pontosság
axs[0, 0].plot(samples, top1_acc, marker='o', label='Top-1 Accuracy')
axs[0, 0].plot(samples, top5_acc, marker='s', label='Top-5 Accuracy')
axs[0, 0].set_title('Top-1 és Top-5 Accuracy')
axs[0, 0].legend()
axs[0, 0].grid(True)

# Precision, Recall, F1-score
axs[0, 1].plot(samples, precision, marker='o', label='Precision')
axs[0, 1].plot(samples, recall, marker='s', label='Recall')
axs[0, 1].plot(samples, f1_score, marker='^', label='F1-score')
axs[0, 1].set_title('Precision, Recall, F1-score')
axs[0, 1].legend()
axs[0, 1].grid(True)

# Inference time
axs[1, 0].plot(samples, inference_time, marker='o', color='red')
axs[1, 0].set_title('Inference Time')
axs[1, 0].set_ylabel('Time (s)')
axs[1, 0].grid(True)

# Üres subplot eltüntetése
fig.delaxes(axs[1, 1])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Adatok betöltése
df = pd.DataFrame(data)

sample_order = ['1k', '5k', '10k', '25k']
df['Samples'] = pd.Categorical(df['Samples'], categories=sample_order, ordered=True)

# Kiválasztjuk a 25k mintás adatokat
df_25k = df[df['Samples'] == '25k']

# Radar chart metrikák
metrics = ['Top-1 Accuracy', 'Top-5 Accuracy', 'Precision', 'Recall', 'F1-score']

# Átalakítjuk az adatokat
labels = df_25k['Model'].values
values = df_25k[metrics].values

# Radar charthoz előkészítés
num_vars = len(metrics)
angles = np.linspace(0, 2 * np.pi, num_vars, endpoint=False).tolist()
angles += angles[:1]  # Körbeérünk

# Plot
fig, ax = plt.subplots(figsize=(10, 10), subplot_kw=dict(polar=True))

for i, model_values in enumerate(values):
    stats = model_values.tolist()
    stats += stats[:1]  # Körbezárás
    ax.plot(angles, stats, label=labels[i])
    ax.fill(angles, stats, alpha=0.1)

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_thetagrids(np.degrees(angles[:-1]), metrics)
plt.legend(loc='best', bbox_to_anchor=(1.1, 1.05))
plt.title('Modellek összehasonlítása - 25k mintával', size=15)
plt.show()
```
