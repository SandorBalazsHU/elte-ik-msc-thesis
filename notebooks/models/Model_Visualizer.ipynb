```python
# Telepítések (ha szükséges Colab környezetben)
!pip install torchview --quiet
!pip install torchvision timm --quiet

# Importálás
import torch
import torchvision.models as models
import timm
from torchview import draw_graph
from PIL import Image

# GPU vagy CPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Modellek betöltése
model_zoo = {
    "ResNet50": models.resnet50(pretrained=True),
    "DenseNet121": models.densenet121(pretrained=True),
    "EfficientNet-B0": timm.create_model("efficientnet_b0", pretrained=True),
    "ConvNeXt-Tiny": timm.create_model("convnext_tiny", pretrained=True),
    "ViT-B/16": timm.create_model("vit_base_patch16_224", pretrained=True),
    "Inception-v3": models.inception_v3(pretrained=True, aux_logits=True),
    "NASNet": timm.create_model("nasnetalarge", pretrained=True),
}

# Dummy input (egységes méret a vizualizáláshoz)
input_tensor = torch.randn(1, 3, 224, 224).to(device)

# Tömör vizualizáció mentése fájlba
for name, model in model_zoo.items():
    model.eval().to(device)
    graph = draw_graph(model, input_data=input_tensor, expand_nested=False, depth=1)
    graph.visual_graph.render(filename=f"{name}_overview", format="png", cleanup=True)
```
