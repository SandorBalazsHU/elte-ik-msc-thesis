## Modellek összehasonlítása

---



```python
import matplotlib.pyplot as plt
import numpy as np

# Metrikák és értékek
metrics = ['Top-1 Accuracy', 'Top-5 Accuracy', 'Precision', 'Recall', 'F1-score']
pretrained = [0.4698, 0.7962, 0.3934, 0.3797, 0.3592]
fine_tuned = [0.7553, 0.9351, 0.7110, 0.6808, 0.6908]

# Oszlopok pozíciója
x = np.arange(len(metrics))
width = 0.35

# Ábra létrehozása
fig, ax = plt.subplots(figsize=(10, 6))
rects1 = ax.bar(x - width/2, pretrained, width, label='Pretrained', color='skyblue')
rects2 = ax.bar(x + width/2, fine_tuned, width, label='Fine-tuned', color='orange')

# Címek és címkék
ax.set_ylabel('Érték')
ax.set_title('ResNet50 teljesítmény összehasonlítása')
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.legend()

# Értékek megjelenítése az oszlopokon
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}',
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3),
                    textcoords="offset points",
                    ha='center', va='bottom')

autolabel(rects1)
autolabel(rects2)

plt.tight_layout()
plt.show()
```
